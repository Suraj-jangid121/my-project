```python
t = (1, 2, 3)

result = t * 3

print("Repeated Tuple:", result)
```

```python
t1 = (1, 2)
t2 = (3, 4)
t3 = (5, 6)

new_tuple = t1 + t2 + t3

print("Joined Tuple:", new_tuple)
```

```python
t = (10, 20, 30, 40, 50)

element = 30

if element in t:
    print("Element exists in the tuple")
else:
    print("Element does not exist in the tuple")
```

```python
numbers = (12, 45, 7, 23, 89, 34)

total = 0
highest = numbers[0]
lowest = numbers[0]

for num in numbers:
    total += num

    if num > highest:
        highest = num

    if num < lowest:
        lowest = num

print("Total:", total)
print("Highest:", highest)
print("Lowest:", lowest)
```

```python
n = (3, 14, 7, 22, 9, 41, 18, 5)

result = ()

for num in n:
    if num > 10:
        result += (num,)

print("Filtered Tuple:", result)
```

```python
s = {"cat", "dog", "bird", "fish"}

count = 0

for item in s:
    count += 1

print("Number of elements:", count)
```

```python
set1 = {1, 2, 3, 4}
set2 = {3, 4, 5, 6}

combined = set1 | set2

print("Combined Set:", combined)
```

```python
s1 = {1, 2, 3, 4}
s2 = {3, 4, 5, 6}

common = s1 & s2

print("Common Elements:", common)
```

```python
s1 = {1, 2, 3, 4}
s2 = {3, 4, 5, 6}

result = s1 ^ s2

print("Elements in either set but not both:", result)
```
